```python
import sqlite3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

sns.set_theme(style="darkgrid")

# Connect to YouTube Analytics Data Warehouse
conn = sqlite3.connect("../database/youtube_trends.db")
df = pd.read_sql("SELECT * FROM video_snapshots;", conn)
conn.close()

print(f"Loaded {len(df):,} verified video snapshots across {df.shape[1]} engineered features.")
df[['title', 'channel_title', 'category_name', 'views', 'likes', 'engagement_rate', 'is_viral']].head(3)
```

## 1. High-Level Portfolio Metrics & Central Tendencies
Analyzing median vs mean to account for extreme viral power-law distributions.

```python
print(f"Total Analyzed Videos:       {len(df):,}")
print(f"Median View Count:          {df['views'].median():,.0f}")
print(f"Mean View Count:            {df['views'].mean():,.0f}")
print(f"Median Engagement Rate:     {df['engagement_rate'].median():.2f}%")
print(f"Viral Videos (>= 1M views): {df['is_viral'].sum():,} ({df['is_viral'].mean()*100:.1f}%)")
```

## 2. One-Way ANOVA: Category Engagement Variance
Testing whether audience engagement rates differ with statistical significance across content categories.

```python
categories_list = [group['engagement_rate'].values for _, group in df.groupby('category_name')]
f_stat, p_val = stats.f_oneway(*categories_list)

print(f"One-Way ANOVA F-Statistic: {f_stat:.2f}")
print(f"p-value:                  {p_val:.2e}")
print(f"Statistically Significant: {p_val < 0.001}")
```

## 3. Algorithmic Pickup Velocity
Evaluating the days elapsed from upload to reaching the trending page.

```python
within_48h = (df['days_to_trend'] <= 2).mean() * 100
within_72h = (df['days_to_trend'] <= 3).mean() * 100
median_days = df['days_to_trend'].median()

print(f"Trending within 48 Hours: {within_48h:.2f}%")
print(f"Trending within 72 Hours: {within_72h:.2f}%")
print(f"Median Days to Trend:    {median_days:.1f} days")
```
